<div class="alert alert-success">
     9.   a) ¿Cuál es el día de más ganancias por cada mes?     b) ¿Cuál es el mes con más ganancias y el de menos ganancias?
    
</div> 

````python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ganancia'] = df['Quantity Ordered'] * df['Price Each']


df['Dia'] = df['Order Date'].dt.date
df['Mes'] = df['Order Date'].dt.month

# Día de más ganancias por cada mes
ganancias_por_dia = df.groupby('Dia')['Ganancia'].sum().reset_index()
ganancias_por_dia['Mes'] = pd.to_datetime(ganancias_por_dia['Dia']).dt.month

dia_max_por_mes = ganancias_por_dia.loc[ganancias_por_dia.groupby('Mes')['Ganancia'].idxmax()]

# Ordenar de mayor a menor ganancias
dia_max_por_mes = dia_max_por_mes.sort_values(by='Ganancia', ascending=False)

print("Día de más ganancias por cada mes (ordenado de mayor a menor):")
print(dia_max_por_mes)

# Mes con más y menos ganancias
ganancias_por_mes = df.groupby('Mes')['Ganancia'].sum()

mes_max = ganancias_por_mes.idxmax()
mes_min = ganancias_por_mes.idxmin()

print(f"\nEl mes con más ganancias es: {mes_max} con {ganancias_por_mes[mes_max]:,.2f}")
print(f"El mes con menos ganancias es: {mes_min} con {ganancias_por_mes[mes_min]:,.2f}")
````
