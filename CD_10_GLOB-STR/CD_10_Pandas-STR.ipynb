#  Lección: Uso de `.str` en Pandas

En Pandas  `.str` permite aplicar operaciones de cadenas de texto **de manera vectorizada** sobre toda una columna (Serie).

Esto evita tener que recorrer fila por fila con bucles `for` y hace el código mucho más eficiente.


## 1. Convertir texto a mayúsculas y minúsculas

Podemos usar `.str.upper()` y `.str.lower()` para normalizar el texto.

```python
df = pd.DataFrame({"nombre": ["Ana", "luis", "MARÍA"]})
df
```

```python

df["nombre_mayus"] = df["nombre"].str.upper()
df["nombre_minus"] = df["nombre"].str.lower()
df
```

##  2. Eliminar espacios en blanco
Con `.str.strip()` quitamos espacios al inicio y final del texto.



```python
df = pd.DataFrame({"palabras": ["  hola ", " mundo  ", " pandas "]})
df
```

```python
df["palabras_limpias"] = df["palabras"].str.strip()
df
```

##  3. Reemplazar texto
Con `.str.replace()` sustituimos un texto por otro.

```python
df = pd.DataFrame({"frase": ["hola mundo", "pandas es genial"]})
df
```

```python
df["frase_mod"] = df["frase"].str.replace("mundo", "Python")
df
```

##  4. Extraer texto usando expresiones regulares (regex)
Con `.str.extract()` podemos obtener partes de un texto.
Ejemplo: extraer el **dominio** de un correo electrónico.

* https://docs.python.org/es/3/howto/regex.html

```python
df = pd.DataFrame({"email": ["ana@mail.com", "luis@uni.mx", "maria@empresa.org"]})
df
```

```python
df["dominio"] = df["email"].str.extract(r'@(.+)$')  # lo que está después del @
df
```

## 5. Dividir texto en columnas
Con `.str.split()` podemos separar un string en partes.  

```python
df = pd.DataFrame({"email": ["ana@mail.com", "luis@uni.mx", "maria@empresa.org"]})
df
```

```python
df["email"].str.split('@') 
```

```python
df["email"].str.split('.') 
```

```python
df["dominio"] = df["email"].str.split('@').str[1]  # lo que está después del @
df
```

##  6. Buscar si un texto contiene una palabra
Con `.str.contains()` podemos verificar si un texto existe dentro de la cadena.


```python
df = pd.DataFrame({"texto": ["Tres tristes tigres", "tragaban trigo en un trigal",
                              "en tres tristes trastos", "tragaban trigo","treS tristes tigres"]})
df
```

```python
df["texto"].str.contains("tres")
```

```python
df["texto"].str.contains("tres", case=False) #case sirve para ignorar mayúsculas/minúsculas
```

```python
df[ df["texto"].str.contains("tres", case=False)
```
