Cada transformador es una tupla de tres elementos que define el nombre del transformador, la transformación a aplicar y los índices de columna a los que aplicar. Por ejemplo: **(Nombre, objeto, columnas)**

## 6 .  Afina(fine tune) el  modelo


# Hyperparámetros clave para **Redes Neuronales** (Multi-Layer Perceptron)

El modelo **`MLPRegressor`** (Multi-Layer Perceptron Regressor) implementa una red neuronal completamente conectada (feedforward) que puede aprender relaciones lineales y no lineales entre las variables de entrada y salida.  
Es muy sensible al **escalado de datos**, por lo que **debe usarse junto con `StandardScaler()`**.

---

## 1. `hidden_layer_sizes`
- **Qué es:** define la arquitectura de la red, es decir, el número de neuronas en cada capa oculta.
- **Efecto:** más capas o neuronas → mayor capacidad de aprendizaje, pero más riesgo de sobreajuste.
- **Probar:**  
  - `(50,)` → 1 capa oculta con 50 neuronas  
  - `(100,)` → 1 capa con 100 neuronas  
  - `(50, 50)` → 2 capas con 50 neuronas cada una  
  - `(100, 50, 25)` → 3 capas (más compleja)

---

## 2. `activation`
- **Qué es:** función de activación de las neuronas.
- **Efecto:** introduce no linealidad al modelo.
- **Probar:**
  - `"relu"` → rápida y estable, la más común.  
  - `"tanh"` → útil en relaciones suaves y centradas en cero.  
  - `"logistic"` → similar a sigmoide (menos usada).  
  - `"identity"` → lineal (sin activación).

---

## 3. `solver`
- **Qué es:** algoritmo de optimización para el entrenamiento.
- **Efecto:** afecta velocidad y convergencia del entrenamiento.
- **Probar:**
  - `"adam"` → optimizador general recomendado (default).  
  - `"lbfgs"` → eficiente en datasets pequeños y convergencia rápida.  
  - `"sgd"` → más control (permite momentum, learning rate manual).

---

## 4. `alpha`
- **Qué es:** término de regularización L2 (penaliza pesos grandes).
- **Efecto:** controla el sobreajuste.
- **Probar:** `0.0001, 0.001, 0.01, 0.1`

---

## 5. `learning_rate` y `learning_rate_init`
- **Qué es:** controla cómo se ajustan los pesos en cada iteración.
- **Efecto:**  
  - Demasiado alto → no converge.  
  - Demasiado bajo → aprendizaje lento.
- **Probar:**
  - `learning_rate` → `"constant"`, `"adaptive"`, `"invscaling"`  
  - `learning_rate_init` → `0.001, 0.01, 0.05`

---

## 6. `max_iter`
- **Qué es:** número máximo de iteraciones (épocas) para entrenar la red.
- **Efecto:** más iteraciones permiten convergencia, pero tardan más.
- **Probar:** `200, 500, 1000`

---

## 7. `batch_size`
- **Qué es:** cantidad de muestras usadas por iteración de actualización de pesos.
- **Efecto:**  
  - Pequeños lotes → entrenamiento más ruidoso pero mejor generalización.  
  - Lotes grandes → más estable pero puede atascarse.
- **Probar:** `16, 32, 64, 128`

---

## 8. `early_stopping`
- **Qué es:** detiene el entrenamiento si no mejora el error de validación tras varias iteraciones.
- **Efecto:** evita sobreentrenamiento automáticamente.
- **Probar:** `True` (recomendado), `False`


```python
#Aplicar el modelo MLPRegressor
from sklearn.neural_network import MLPRegressor
```

```python
#1. Nombrar una variable para nuestro modelo
MLP_regression = MLPRegressor()
```

```python
MLP_regression.get_params()
```

```python
#2. Definir el Pipeline : Transformación  + modelo
modelo_MLP= Pipeline( [  ('transformacion', transformacion   ) ,
                                     ('MLP_regression',MLP_regression) ]  )
```

```python
# Utilizamos la función GridSearchCV para encontrar los valores optimos
from sklearn.model_selection import GridSearchCV
```

```python
#Definición de posibles parametros
hidden_layer_sizes = [(50,),   (50, 50) ,(100,) ]
activation = ['relu', 'tanh']
learning_rate_init = [0.001,   0.05]
solver = ['adam', 'lbfgs']

# Diccionario de parámetros con el prefijo del pipeline (por ejemplo: 'MLP_regression')
parametros = {
'MLP_regression__hidden_layer_sizes': hidden_layer_sizes,
'MLP_regression__activation': activation,
'MLP_regression__learning_rate_init': learning_rate_init,
'MLP_regression__solver': solver
}
```

```python
#Definicion del grid de busqueda
grid_MLP = GridSearchCV( estimator=modelo_MLP,
                        param_grid=parametros,
                        cv=5,
                        n_jobs=-1,
                        scoring='neg_mean_squared_error' ,
                        verbose=2)
```

```python
# Ajuste del modelo usando todas las posibles combinaciones de parametros 
grid_MLP.fit(X,y)
```

```python
# Mostrar la mejor combinación
print( grid_MLP.best_params_ )
```

**Encontramos los nuevos valores de la media y desviación estándar**


```python
#1 Ajustar el modelo con los nuevos parametros 
MLP_regression_tunned = MLPRegressor(activation='tanh', hidden_layer_sizes= (50, 50), learning_rate_init=0.001, solver='adam')
```

```python
MLP_regression_tunned.get_params()
```

```python
#2. Definir el Pipeline : Transformación  + modelo
modelo_MLP_tunned = Pipeline( [  ('transformacion', transformacion   ) ,
                                     ('MLP_regression_tunned',MLP_regression_tunned) ]  )
```

```python
#3. Ajustar el modelo
modelo_MLP_tunned.fit(X,y)
```

```python
from sklearn.model_selection import cross_val_score

# 4. Calcular el score con la validación cruzada CV=5
cv_MLP_tunned = cross_val_score(modelo_MLP_tunned, X, y, cv=5, scoring='neg_mean_squared_error')
cv_MLP_tunned
```

```python
# 5. Realizar el cambio de signo y sacar la raiz
error_MLP_tunned = np.sqrt(-cv_MLP_tunned)
error_MLP_tunned
```

```python
# 6. Mostrar el resultado usando la funcion display_scores
media_error_MLP_tunned = np.mean(error_MLP_tunned)
std_error_MLP_tunned = np.std(error_MLP_tunned)
print("Media :", media_error_MLP_tunned)
print("Desviación estándar :", std_error_MLP_tunned)
```
