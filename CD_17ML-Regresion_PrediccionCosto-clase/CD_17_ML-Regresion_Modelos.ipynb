```python
import joblib
import glob
import os
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# 1. Carga tus datos de prueba
df=pd.read_csv("datos/USA_Housing_Test.txt")
df.dropna(inplace=True)
df.head(2)
```

```python
X_test =df.drop("Price", axis=1)
X_test.head(2)
```

```python
y_test=df[ "Price"]
y_test.head(2)
```

```python
#modelos
ruta_modelos = "modelos/"  # cámbiala a tu carpeta
archivos_modelos = glob.glob(os.path.join(ruta_modelos, "*.joblib"))
```

```python
resultados = []
for modelo_path in archivos_modelos:
    # nombre del modelo a partir del archivo
    nombre_modelo = os.path.basename(modelo_path).replace(".joblib", "")
    
    # 2. cargar modelo
    modelo = joblib.load(modelo_path)
    
    # 3. predecir
    y_pred = modelo.predict(X_test)
    
    # 4. métricas
    r2 = r2_score(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    
    resultados.append({
        "modelo": nombre_modelo,
        "r2": r2,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })

# 5. Pasar a DataFrame para comparar
df_resultados = pd.DataFrame(resultados).sort_values(by="r2", ascending=False)
print(df_resultados)
```

```python
# Cargar el modelo
modelo = joblib.load("modelos/MEJOR-MODELO. joblib")

# Hacer las predicciones
y_pred = modelo.predict(X_test)

# Guardar las predicciones en un archivo CSV
pd.DataFrame(y_pred, columns=["Prediccion"]).to_csv("predicciones_ENRIQUECAMACHO.csv", index=False)

print("Archivo guardado correctamente.")
```
