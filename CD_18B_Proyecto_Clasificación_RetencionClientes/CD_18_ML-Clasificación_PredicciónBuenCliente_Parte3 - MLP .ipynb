## 1. Obtención y visualización de los datos

```python
df = pd.read_csv('datos/Datos_Train-Validation.csv')
df
```

## 2. Preparación de los datos y funciones

```python
# Preparación de datos
from sklearn.model_selection import train_test_split
```

#### División general de los datos


* **X**: Datos sin el diagnostico es decir sin la columna "target"
* **y**: Datos con el diagnostico es decir con la columna "target"

Muestras de entrenamiento y prueba
* **X_train**: este será el grupo de entrenamiento
* **X_test**: este será el  grupo de prueba
* **y_train**: este será el target para el grupo de entrenamiento
* **y_test**: este será el target para el grupo de prueba

```python
#Crear X (Quitar la columna o campo 24)
X =df.drop(['id','CustomerId','Surname','Exited'], axis=1)
X.head(2)
```

```python
# Crear y (Valor de de la columna "target")
y=df['Exited']
y.head()
```

```python
X
```

```python
# Dividir los datos para entrenamiento y prueba. Con un factor 80%
X_train , X_test, y_train, y_test = train_test_split(X, y, train_size=0.8)
print( 'X_train :' , X_train.shape )
print( 'X_test :' , X_test.shape )
print( 'y_train :' , y_train.shape )
print( 'y_test :' ,  y_test.shape )
print( 'df 100% :' ,  df.shape[0]   )
print( 'df 80% :' ,  df.shape[0] * 0.8 )
```

### Pipeline

<img src="figuras/pipeline.jpg" align="middle" style="width:400px;" />

```python
# Modulos importantes
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
```

```python
# Mostrando columnas
X.columns
```

```python
# Crear la lista de  viriables numericas  numeric_list 
numeric_list = ['CreditScore', 'Gender-0Male-1Female', 'Age', 'Tenure', 'Balance',
       'NumOfProducts', 'HasCrCard', 'IsActiveMember', 'EstimatedSalary']
```

```python
# Crear el pipeline de preprocesamiento para datos numéricos  numeric_transformer
numeric_transformer = Pipeline(  steps = [ ('scaler',  StandardScaler()  )  ]   )
```

```python
# Definir los transformadores
transformer = ColumnTransformer( [ ('num', numeric_transformer, numeric_list  )  ]  )
```

# 3. Implementación de Modelos

## Segundo modelo: Multi Layer Perceptron (MLP)

Referencia
* https://michael-fuchs-python.netlify.app/2021/02/03/nn-multi-layer-perceptron-classifier-mlpclassifier/

## a) MLP sin tunning

```python
# Módulos del Modelo
from sklearn.neural_network import MLPClassifier

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
# modelo DecisionTreeClassifier llamado DT_Classifier
MLP_Classifier = MLPClassifier(max_iter = 1000)
```

```python
# Parametros del modelo DecisionTreeClassifier()
MLP_Classifier.get_params()
```

```python
# Definiendo el modelo_DT con su pipeline
modelo_MLP = Pipeline(  [ ('transformer', transformer) , 
                        ('MLP_Classifier',MLP_Classifier)    ]  )
```

```python
# Ajuste del modelo
modelo_MLP.fit(X_train, y_train)
```

```python
# Score sin tunning
modelo_MLP.score(X_test, y_test)
```

## b) MLP con tunning

```python
# Definir los parametros con el diccionario params utilizando:
hidden_layer_sizes= [(150,100,50), (120,80,40), (100,50,30)]
max_iter= [ 500 ]
activation= ['tanh', 'relu']
solver= ['sgd']
alpha= [0.0001, 0.05]

params = { 'MLP_Classifier__hidden_layer_sizes' : hidden_layer_sizes,
         'MLP_Classifier__max_iter': max_iter,
         'MLP_Classifier__activation': activation ,
         'MLP_Classifier__solver': solver ,
         'MLP_Classifier__alpha': alpha }
```

```python
# Ajuste de los parametros max_depth y min_samples_leaf con  GridSearchCV
from sklearn.model_selection import GridSearchCV

grid_MLP = GridSearchCV(modelo_MLP, 
                       param_grid= params,
                      cv=5,
                      scoring='neg_mean_squared_error',
                      n_jobs=-1,
                      verbose=10)
```

```python
# Ajuste del modelo usando todas las posibles combinaciones de parametros 
grid_MLP.fit(X, y)
```

```python
#Encontrar los mejores parametros con best_params_
grid_MLP.best_params_
```

```python
# Ajustar el modelo con los nuevos parametros nombrandolo como  DT_Classifier_tunned
MLP_Classifier_tunned =  MLPClassifier(activation='tanh',
                                              alpha=0.0001 , 
                                                hidden_layer_sizes=(150, 100, 50),
                                               max_iter=500,
                                               solver='sgd')
```

```python
# Definiendo el modelo_DT con su pipeline
modelo_MLP_tunned = Pipeline(  [ ('transformer', transformer) , 
                        ('MLP_Classifier_tunned',MLP_Classifier_tunned)    ]  )
```

```python
#  Ajustar el modelo
modelo_MLP_tunned.fit(X_train, y_train)
```

```python
# Predicción
y_predict = modelo_MLP_tunned.predict(X_test)
y_predict
```

```python
# Prueba en y_test
y_test.values
```

```python
# Comparación
#np.equal(  y_predict , y_test.values ) 
```

```python
# Score
np.equal(  y_predict , y_test.values ).sum() / len(y_predict)
```

```python
# Score
modelo_MLP_tunned.score(X_test, y_test)
```

```python
#Evaluacion
# Calcular el score con la validación cruzada CV=5
from sklearn.model_selection import cross_val_score

MLP_tunned_scores= cross_val_score( modelo_MLP_tunned, X, y,
                              scoring= 'neg_mean_squared_error', cv=5)
MLP_tunned_scores
```

```python
# Realizar el cambio de signo y sacar la raiz
MLP_tunned_rmse_scores= np.sqrt(-MLP_tunned_scores)
MLP_tunned_rmse_scores
```

```python
#Calcular mean y standard deviation
media_MLP_tunned= MLP_tunned_rmse_scores.mean() 
std_MLP_tunned=MLP_tunned_rmse_scores.std()
print( media_MLP_tunned)
print( std_MLP_tunned)
```
