# Ejercicio 3: Pacientes confirmados con Covid en México


Fuente: https://datos.covid-19.conacyt.mx/#DownZCSV

```python
df= pd.read_csv("datos/Casos_Diarios_Estado_Nacional_Confirmados_20201129.csv")
```

```python
df.head()
```

<div class="alert alert-success">
    
1. Hacer una grafica de barras comparando todos los meses, considerando el total nacional (por mes) 
    
</div> 

```python
# Estilo Matlab
# Fila nacional
df_nac = df.loc[df['nombre'] == 'Nacional']

# Convertir columnas de fechas a serie con índice datetime
casos = df_nac.iloc[0, 3:].rename('casos')
casos.index = pd.to_datetime(casos.index, format='%d-%m-%Y')

# Agrupar por mes y sumar
casos_mensuales = casos.resample('ME').sum()

# Gráfica
plt.figure(figsize=(10,5))
plt.bar(casos_mensuales.index.strftime('%B'), casos_mensuales.values, color='red', edgecolor='black')
plt.xlabel('Mes')
plt.ylabel('Total de casos confirmados')
plt.title('Casos confirmados')
plt.xticks(rotation=35)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
#Orientado a Objeto

# Fila nacional
df_nac = df.loc[df['nombre'] == 'Nacional']

# Convertir columnas de fechas a serie con índice datetime
casos = df_nac.iloc[0, 3:].rename('casos')
casos.index = pd.to_datetime(casos.index, format='%d-%m-%Y')

# Agrupar por mes y sumar
casos_mensuales = casos.resample('ME').sum()

fig, ax = plt.subplots(figsize=(10,5), dpi=100)
ax.bar(casos_mensuales.index.strftime('%B'), casos_mensuales.values, color='red', edgecolor='black')

ax.set_xlabel('Mes')
ax.set_ylabel('Total de casos confirmados')
ax.set_title('Casos confirmados')
ax.set_xticklabels(casos_mensuales.index.strftime('%B'), rotation=35)
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

<div class="alert alert-success">
    
2. Hacer una grafica de barras comparando  los meses de agosto, septiembre, octubre y noviembre, considerando solo las entidades de Yucatán, Distrito Federal  y Jalisco
    
</div> 

```python
# Estilo Matlab
entidades = ['YUCATAN', 'DISTRITO FEDERAL', 'JALISCO']
meses_num = [8, 9, 10, 11]
meses_es = ['Agosto', 'Septiembre', 'Octubre', 'Noviembre']

df_sel = df[df['nombre'].isin(entidades)]
fechas = df_sel.columns[3:]
fechas_dt = pd.to_datetime(fechas, format='%d-%m-%Y')

resultados = []
for entidad in entidades:
    casos_ent = df_sel[df_sel['nombre'] == entidad][fechas].reset_index(drop=True)
    suma_mes = []
    for mes in meses_num:
        idx_mes = [i for i, f in enumerate(fechas_dt) if f.month == mes]
        if not casos_ent.empty and len(idx_mes) > 0:
            suma = casos_ent.iloc[:, idx_mes].sum(axis=1).values[0]
        else:
            suma = 0
        suma_mes.append(suma)
    resultados.append(suma_mes)

plt.figure(figsize=(10,5))
width = 0.2
x = np.arange(len(meses_es))
for i, entidad in enumerate(entidades):
    plt.bar(x + i*width, resultados[i], width, label=entidad)

plt.xlabel('Mes')
plt.ylabel('Total de casos confirmados')
plt.title('Casos confirmados por mes y entidad')
plt.xticks(x + width, meses_es, rotation=35)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()


```
