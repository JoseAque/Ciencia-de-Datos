## Ejercicio 1:  Datos de la estación meteorologica de Progreso, Yucatán

Fuente:  https://smn.conagua.gob.mx/es/observando-el-tiempo/estaciones-meteorologicas-automaticas-ema-s

```python
df= pd.read_csv("datos/Estacion_PROGRESO_90_dias_19Nov.csv")
df.head()
```

```python
df.shape
```

<div class="alert alert-success">

1. Considerando todos los datos hacer una grafica de dispersión para la Temperatura de Aire (en el eje Y) y la Humadad Relativa (en el eje X) 
</ul>
</div>

```python
# Estilo Matlab
plt.figure(figsize=(6,3))  

plt.scatter(df['HumedadRelativa(%)'], df['TemperaturaAire(C)'], c= 'blue' , s = 10 , marker='X')
plt.title('Temperatura de Aire - Humedad Relativa',fontsize=10)
plt.xlabel('Humedad Relativa (%)',fontsize=10)
plt.ylabel('Temperatura de Aire (C)',fontsize=10)

plt.show()
```

```python
# Estilo Orientado a Objetos
fig, ax = plt.subplots(figsize=(6,3), dpi=100)

ax.scatter(df['HumedadRelativa(%)'], df['TemperaturaAire(C)'], c= 'blue' , s = 10 , marker='X')
ax.set_title('Temperatura de Aire - Humedad Relativa',fontsize=10)
ax.set_xlabel('Humedad Relativa (%)',fontsize=10)
ax.set_ylabel('Temperatura de Aire (C)',fontsize=10)

plt.show()
```

<div class="alert alert-success">
    
2. Considerando todos los datos, hacer un histograma de los valores de la Rapidez Viento 
    
</div> 

```python
# Estilo Matlab
plt.figure(figsize=(6,5))  

plt.hist(df['RapidezViento(km/h)'], 110, density=False, facecolor='red', edgecolor = 'darkred')

plt.xlabel('RapidezViento(km/h)')
plt.ylabel('Frecuencia')
plt.title('Rapidez del Viento')

plt.grid(True, linestyle = 'solid', alpha = 0.5)
plt.show()
```

```python
# Estilo Orientado a Obejetos
fig, ax = plt.subplots(figsize=(6,5), dpi=100)

ax.hist(df['RapidezViento(km/h)'], 110, density=False, facecolor='red', edgecolor = 'darkred')
ax.set(title='Rapidez del Viento', xlabel= 'RapidezViento(km/h)', ylabel='Rapidez del Viento')
ax.grid(True, linestyle = 'solid', alpha = 0.5)

plt.show()
```

# Ejercicio 2: Incendios en México

```python
df= pd.read_csv("datos/Serie_historica_anual_incendios_2017.csv")
df.head()
```

```python
df.columns
```

<div class="alert alert-success">
    
1. Utilizar una grafica de barras para comparar el total de hectáreas afectadas para los 5 estados más afectados 
    
</div> 

```python
df5 = pd.DataFrame(df.groupby('Estado')['Total Hectareas'].sum().sort_values(ascending = False).head(5).reset_index())
df5
```

```python
#Estilo Matlab
plt.figure(figsize=(6,3))  

plt.bar(df5['Estado'], df5['Total Hectareas'], facecolor='lightgreen', edgecolor = 'darkred')          

plt.title('5 estados más afectados',fontsize=10)
plt.xlabel('Hectareas',fontsize=10)
plt.ylabel('Frecuencia',fontsize=10);

plt.show()
```

```python
# Estilo Orientado a Obejetos
fig, ax = plt.subplots(figsize=(6,3), dpi=100)

ax.bar(df5['Estado'], df5['Total Hectareas'], facecolor='lightgreen', edgecolor = 'darkred')      
ax.set(title='Total de emisiones de C02', xlabel= 'Paises' , ylabel='Total de Emisiones de C02 ')

plt.show()
```

<div class="alert alert-success">
    
2. Utilizar una grafica de barras para comparar el total de hectáreas afectadas para Yucatán para los meses: abril, mayo, junio, julio y agosto 
    
</div> 

```python
df_y = df[df['Estado']=='Yucatán']
```

```python
df_y = pd.DataFrame(df_y.groupby('Mes Acontecimiento')['Total Hectareas'].sum().reset_index())
df_y.head(12)
```

```python
#Estilo Matlab
plt.figure(figsize=(6,3))  

plt.bar(df_y['Mes Acontecimiento'], df_y['Total Hectareas'], facecolor='lightblue', edgecolor = 'blue')          

plt.title('Areas Afectadas en Yucatan',fontsize=10)
plt.xlabel('Hectareas',fontsize=10)
plt.ylabel('Frecuencia',fontsize=10);

plt.show()
```

```python
# Estilo Orientado a Obejetos
fig, ax = plt.subplots(figsize=(6,3), dpi=100)

ax.bar(df_y['Mes Acontecimiento'], df_y['Total Hectareas'], facecolor='lightblue', edgecolor = 'blue')      
ax.set(title='Total de emisiones de C02', xlabel= 'Paises' , ylabel='Total de Emisiones de C02 ')

plt.show()
```
